# Softmax Regression Implementation

Just as we implemented linear regression both from scratch and concisely in that section, we now build softmax regression on Fashion-MNIST both ways. We start from raw tensors—implementing the rowwise softmax, vectorized cross-entropy loss, and `SoftmaxRegressionScratch` parameters ($\mathbf{W} \in \mathbb{R}^{784 \times 10}, \mathbf{b} \in \mathbb{R}^{10}$) alongside the reusable `Classifier(d2l.Module)` base class—train to roughly $83\%$ validation accuracy, and inspect what the linear model learned through its $10 \times 10$ confusion matrix and $28 \times 28$ weight templates. In that section, we then replace manual tensors and probability clamping with a built-in linear layer and a shifted *log-sum-exp* cross-entropy loss computed directly from raw logits, eliminating floating-point overflow and underflow at once.

```python
import math
import numpy as np
import torch
from torch import nn
from torch.nn import functional as F
from d2l import torch as d2l
```

<!-- d2l:prose id=softmax-regression-scratch-md-2 fw=all -->

## The Softmax

### Normalizing Each Row

For a minibatch of $n$ examples, the logits form an $n \times q$ matrix $\mathbf{O} = \mathbf{X}\mathbf{W} + \mathbf{b}$ (that section). Softmax normalizes each row independently:

$$\mathrm{softmax}(\mathbf{O})_{ij} = \frac{\exp(O_{ij})}{\sum_{k=1}^q \exp(O_{ik})}.$$

In code, we exponentiate `O`, sum across columns (`axis=1`) with `keepdims=True` so the row sums keep shape `(n, 1)`, and divide via broadcasting:

```python
def softmax(O):
    O_exp = d2l.exp(O)
    return O_exp / d2l.reduce_sum(O_exp, 1, keepdims=True)  # Broadcast
```

<!-- d2l:prose id=softmax-regression-scratch-md-3 fw=all -->

Let us sanity-check `softmax` on two rows: the cat/chicken/dog logits
$(1.0, 2.2, 0.3)$ from that section, and a row of three identical
zeros.

```python
O = d2l.tensor([[1.0, 2.2, 0.3], [0.0, 0.0, 0.0]])
Y_hat = softmax(O)
Y_hat, d2l.reduce_sum(Y_hat, 1)
```

<!-- d2l:prose id=softmax-regression-scratch-md-4 fw=all -->

### Overflow

Why should you never deploy this three-line `softmax` formula in production? Because in 32-bit floating point (`float32`), numbers top out near $3.4 \times 10^{38}$, so $\exp(o)$ overflows to `inf` as soon as $o > 88.7$. Feed `softmax` the logits $(1000, 0, 0)$ and $\infty / \infty$ evaluates to `NaN`, even though the true probability distribution is $(1, 0, 0)$ to machine precision.

As we proved in that section, softmax is invariant to shifting every logit in a row by the same constant. Subtracting the row maximum $o_{\max} = \max_k o_k$ before exponentiating ensures every exponent $o_j - o_{\max}$ is $\leq 0$ (and the largest is $0$), making overflow impossible. Every framework's built-in softmax applies this shift internally:

```python
O = d2l.tensor([[1000.0, 0.0, 0.0]])
O_max = O.amax(dim=1, keepdim=True)
softmax(O), softmax(O - O_max), torch.softmax(O, dim=1)
```

<!-- d2l:prose id=softmax-regression-scratch-md-5 fw=all -->

For our Fashion-MNIST scratch run below, pixel values lie in $[0, 1]$ and weights stay small, so `softmax` does not overflow during training; however, shifting alone still leaves very negative logits vulnerable to *underflow* ($\exp(-200) = 0.0$ in float32, turning $\log 0$ into $-\infty$). In that section below, we eliminate both hazards permanently by fusing softmax and the logarithm in log space.

## Loss and Accuracy

### Two Uses of the Logits

While gradient descent optimizes the smooth cross-entropy loss, humans judge a classifier by its *accuracy*—the fraction of examples where the top-scoring class matches the true label $y^{(i)}$:

$$\textrm{accuracy} = \frac{1}{n} \sum_{i=1}^n \mathbf{1}\!\Big(\mathop{\mathrm{argmax}}_j o^{(i)}_j = y^{(i)}\Big).$$

Both metrics flow from the exact same logit vector (the figure), and tracking both during validation is essential: if a model becomes overconfident on a few mistakes, its validation cross-entropy loss can rise even while its validation accuracy is still climbing.

![Training (top) maps logits through softmax to the smooth cross-entropy loss; evaluation (bottom) compares $\operatorname{argmax}_j o_j$ against the label $y$ to compute discrete accuracy.](../img/mdl-clf-loss-accuracy.svg)

### The `Classifier` Class

We package this dual-metric validation behavior into `Classifier(d2l.Module)`, the base class for every classification model in the book, starting with `validation_step` returning both `'loss'` and `'acc'` on each validation minibatch:

```python
class Classifier(d2l.Module):
    """The base class of classification models."""
    def validation_step(self, batch):
        Y_hat = self(*batch[:-1])
        return {'loss': self.loss(Y_hat, batch[-1]),
                'acc': self.accuracy(Y_hat, batch[-1])}
```

<!-- d2l:prose id=softmax-regression-scratch-md-6 fw=all -->

Next, we attach `accuracy(self, Y_hat, Y, averaged=True)` to `Classifier`: it flattens `Y_hat` to one row of class scores per example, finds the predicted class index via `argmax` along `axis=1`, casts it to `Y.dtype`, and returns either the mean or per-example float32 equality mask `preds == Y`:

```python
@d2l.add_to_class(Classifier)
def accuracy(self, Y_hat, Y, averaged=True):
    """Compute the fraction of correct predictions."""
    Y_hat = d2l.reshape(Y_hat, (-1, Y_hat.shape[-1]))
    preds = d2l.astype(d2l.argmax(Y_hat, axis=1), Y.dtype)
    compare = d2l.astype(preds == d2l.reshape(Y, (-1,)), d2l.float32)
    return d2l.reduce_mean(compare) if averaged else compare
```

<!-- d2l:prose id=softmax-regression-scratch-md-7 fw=all -->

Finally, we install plain minibatch SGD at learning rate `self.lr` as the default `configure_optimizers` method on `d2l.Module` so every classifier inherits a working optimizer out of the box:

```python
@d2l.add_to_class(d2l.Module)
def configure_optimizers(self):
    return torch.optim.SGD(self.parameters(), lr=self.lr)
```

<!-- d2l:prose id=softmax-regression-scratch-md-8 fw=all -->

## The Model and the Loss

### The Parameters and Forward Pass

Flattening each $28 \times 28$ Fashion-MNIST image yields $d = 784$ input features for $q = 10$ classes, so `SoftmaxRegressionScratch` allocates $\mathbf{W} \in \mathbb{R}^{784 \times 10}$ (initialized from $\mathcal{N}(0, 0.01^2)$) and $\mathbf{b} \in \mathbb{R}^{10}$ (initialized to $0$), totaling $7{,}850$ parameters. Notice a quick sanity check: near-zero initial weights produce near-zero logits, so the untrained model starts close to a uniform prediction ($\hat{y}_j \approx 0.1$) with initial cross-entropy loss near $-\log(0.1) = \log 10 \approx 2.30$:

```python
class SoftmaxRegressionScratch(Classifier):
    """Softmax regression implemented from scratch."""
    def __init__(self, num_inputs, num_outputs, lr, std=0.01):
        super().__init__()
        self.save_hyperparameters()
        self.W = nn.Parameter(d2l.normal(0, std, (num_inputs, num_outputs)))
        self.b = nn.Parameter(d2l.zeros(num_outputs))
```

<!-- d2l:prose id=softmax-regression-scratch-md-9 fw=all -->

In `forward(self, X)`, we flatten each image in the minibatch into a row vector of length `self.num_inputs` ($784$) via `d2l.reshape(X, (-1, self.num_inputs))` and map the affine logits $\mathbf{X}\mathbf{W} + \mathbf{b}$ through `softmax`:

```python
@d2l.add_to_class(SoftmaxRegressionScratch)
def forward(self, X):
    X = d2l.reshape(X, (-1, self.num_inputs))  # One flattened image per row
    return softmax(d2l.matmul(X, self.W) + self.b)
```

<!-- d2l:prose id=softmax-regression-scratch-md-10 fw=all -->

### The Cross-Entropy Loss

Recall from that section that $l(\mathbf{y}, \hat{\mathbf{y}}) = -\log \hat{y}_y$. Instead of materializing one-hot matrices and taking inner products, we pluck out the true-class probability $\hat{y}^{(i)}_{y^{(i)}}$ for every row $i$ in one vectorized indexing step:

```python
y = d2l.tensor([0, 2])
y_hat = d2l.tensor([[0.1, 0.3, 0.6], [0.3, 0.2, 0.5]])
y_hat[[0, 1], y]
```

<!-- d2l:prose id=softmax-regression-scratch-md-11 fw=all -->

That extracts 0.1 for the first example and 0.5 for the second. Wrapping
this in `cross_entropy`, we average the negative logarithms across the
minibatch. Notice the one defensive clamp to $10^{-12}$ before taking the
logarithm: it stops an underflowed zero probability from turning into
$-\infty$, capping the per-example loss at $27.6$, though clamping also
zeroes out the gradient whenever it triggers. Later in this section
(that section), we replace this stopgap with a fused loss computed directly
from logits.

```python
def cross_entropy(y_hat, y):
    # Tiny clip to keep log finite when softmax outputs underflow to 0.
    p = y_hat[list(range(len(y_hat))), y].clamp(min=1e-12)
    return -d2l.reduce_mean(d2l.log(p))

cross_entropy(y_hat, y)
```

<!-- d2l:prose id=softmax-regression-scratch-md-12 fw=all -->

That yields the mean of $-\log 0.1 \approx 2.30$ and $-\log 0.5 \approx 0.69$,
or roughly 1.50. Let us attach it as the model's `loss` method:

```python
@d2l.add_to_class(SoftmaxRegressionScratch)
def loss(self, y_hat, y):
    return cross_entropy(y_hat, y)
```

## Training and Evaluation

### The Training Run

Let us train `SoftmaxRegressionScratch` on Fashion-MNIST for $10$ epochs with minibatch size $256$ and learning rate $\eta = 0.1$:

```python
data = d2l.FashionMNIST(batch_size=256)
model = SoftmaxRegressionScratch(num_inputs=784, num_outputs=10, lr=0.1)
trainer = d2l.Trainer(max_epochs=10)
trainer.fit(model, data)
```

<!-- d2l:prose id=softmax-regression-scratch-md-14 fw=all -->

To score a trained model on an entire dataset split whenever we want, we attach an `evaluate(self, model, data, train=False)` helper to `d2l.Trainer` that binds `self.val_dataloader` to either the validation or training loader and runs `self.validate()`:

```python
@d2l.add_to_class(d2l.Trainer)
def evaluate(self, model, data, train=False):
    """Metrics of `model` on the validation data of `data` (on its training
    data if `train`), averaged over examples."""
    self.train_dataloader = None
    self.val_dataloader = (data.train_dataloader() if train
                           else data.val_dataloader())
    self.prepare_model(model)
    return self.validate()
```

<!-- d2l:prose id=softmax-regression-scratch-md-15 fw=all -->

Averaged across all 10,000 validation images, these metrics match the final
points recorded in `trainer.history['val_loss'][-1]` and
`trainer.history['val_acc'][-1]`.

```python
trainer.evaluate(model, data)
```

<!-- d2l:prose id=softmax-regression-scratch-md-16 fw=all -->

Our classifier gets roughly $83\%$ of the validation images right; passing `train=True` evaluates the exact same weights across all $60{,}000$ training images:

```python
trainer.evaluate(model, data, train=True)
```

<!-- d2l:prose id=softmax-regression-scratch-md-17 fw=all -->

After 10 epochs, validation accuracy settles near $83\%$ (varying by $\pm 1\%$ across random seeds) and training accuracy sits near $84\%$. That tiny $1\%$ generalization gap reflects the restricted capacity of a linear classifier: $7{,}850$ parameters cannot memorize $60{,}000$ images (that section).

### The Confusion Matrix

To summarize the error pattern across all $10{,}000$ validation images, we tally the $10 \times 10$ *confusion matrix* $\mathbf{N} \in \mathbb{R}^{q \times q}$, where rows index the predicted class $\hat{c}^{(m)} = \operatorname{argmax}_j o^{(m)}_j$ and columns index the true class $y^{(m)}$:

$$N_{ij} = \sum_{m=1}^n \mathbf{1}\big(\hat{c}^{(m)} = i\big)\, \mathbf{1}\big(y^{(m)} = j\big).$$

(Placing true classes in the columns and predicted classes in the rows makes column-normalized conditional probabilities multiply cleanly as $\boldsymbol{\mu} = \mathbf{C}\mathbf{p}$ when we study label shift in that section.) Using one-hot matrices $\mathbf{P}, \mathbf{Y} \in \{0, 1\}^{n \times q}$ for a minibatch's predicted and true classes, the product $\mathbf{P}^\top\mathbf{Y}$ accumulates all $100$ counts in a single matrix multiplication per minibatch:

```python
def one_hot(y, q):
    """Rows of the q x q identity matrix, one per class index in y."""
    return d2l.astype(d2l.reshape(y, (-1, 1)) == d2l.arange(q, dtype=y.dtype),
                      d2l.float32)

N = d2l.zeros((10, 10))
for X, y in data.val_dataloader():
    P = one_hot(d2l.argmax(model(X), axis=1), 10)
    N += d2l.matmul(d2l.transpose(P), one_hot(y, 10))
int(d2l.reduce_sum(N))
```

<!-- d2l:prose id=softmax-regression-scratch-md-20 fw=all -->

All 10,000 validation images are accounted for. Because raw counts depend
on class frequency, we normalize each column of $\mathbf{N}$ by its sum to
obtain $C_{ij} = N_{ij} / \sum_{i'} N_{i'j}$, an empirical estimate of the
conditional probability $P(\hat{c} = i \mid y = j)$. Each column shows
where examples of true class $j$ land, while the diagonal gives per-class
recognition rates. Let us render $\mathbf{C}$ as a heatmap via
`d2l.show_heatmaps`.

```python
C = N / d2l.reduce_sum(N, 0, keepdims=True)  # Each column sums to 1
names = data.text_labels(range(10))
d2l.show_heatmaps(d2l.reshape(C, (1, 1, 10, 10)), xlabel='true class',
                  ylabel='predicted class', figsize=(4, 4), cmap='Blues',
                  xticklabels=names, yticklabels=names)
```

<!-- d2l:prose id=softmax-regression-scratch-md-21 fw=all -->

Let us also print the five largest off-diagonal entries to spotlight the
most frequent confusions:

```python
mistakes = d2l.numpy(N) * (1 - np.eye(10))  # Zero the diagonal
for k in np.argsort(mistakes, axis=None)[::-1][:5]:
    i, j = divmod(int(k), 10)
    print(f'{names[j]} taken for {names[i]}: {int(mistakes[i, j])}')
```

<!-- d2l:prose id=softmax-regression-scratch-md-22 fw=all -->

The confusion matrix confirms what our visual inspection in that section predicted: trousers, bags, and footwear are recognized $\sim 90\%$ of the time or better, whereas the five largest off-diagonal confusions all occur among the four look-alike upper-body tops (T-shirt, pullover, coat, and shirt)—with shirts recognized only about half the time or less because their silhouette sits right in the middle of the other three.

### The Learned Templates

Because each logit $o_j = \mathbf{w}_j^\top\mathbf{x} + b_j$ is an inner product between the flattened image $\mathbf{x}$ and column $j$ of $\mathbf{W}$, reshaping each 784-element column $\mathbf{w}_j$ back to $28 \times 28$ lets us view the entire classifier as ten visual *templates* (red = positive weights, blue = negative weights):

```python
W = d2l.numpy(model.W)
templates = W.T.reshape(10, 28, 28)  # Column j of W as an image
v = np.abs(templates).max()
d2l.show_images(templates, 2, 5, titles=names, cmap='RdBu_r', vmin=-v, vmax=v);
```

<!-- d2l:prose id=softmax-regression-scratch-md-23 fw=all -->

Each template is a recognizable ghostly silhouette of its garment—notice the blue negative channel between the trouser legs that penalizes dresses, and the similar torso shapes of the four tops that differ mainly around the collar and sleeves. Because a linear layer can only take a single fixed global dot product per class, subtle local features like collars or buttons cannot overcome overall silhouette overlap; breaking past $\sim 83\%$ requires nonlinear hidden layers (that section) and spatial convolutions (that section).

## Concise Implementation

Now let us see how deep learning frameworks simplify softmax regression while permanently curing the overflow and underflow hazards of our naive `softmax` and clamped `cross_entropy`.

### Defining the Model

In `SoftmaxRegression(d2l.Classifier)`, a flattening step and a single built-in fully connected layer with $10$ outputs compute the logits $\mathbf{O} = \mathbf{X}\mathbf{W} + \mathbf{b}$ (initialized with $\mathcal{N}(0, 0.01^2)$ weights and zero biases to match our scratch baseline). Crucially, **`forward` returns the raw logits $\mathbf{O}$ directly, without applying softmax**:

```python
class SoftmaxRegression(d2l.Classifier):
    """The softmax regression model."""
    def __init__(self, num_outputs, lr):
        super().__init__()
        self.save_hyperparameters()
        self.net = nn.Sequential(nn.Flatten(),
                                 nn.LazyLinear(num_outputs))

    def weight_init(self, module):
        # The Trainer calls this after the lazy layer has learned its shape
        if isinstance(module, nn.Linear):
            nn.init.normal_(module.weight, std=0.01)
            nn.init.zeros_(module.bias)

    def forward(self, X):
        return self.net(X)
```

<!-- d2l:prose id=softmax-regression-scratch-md-24 fw=all -->

### A Stable Cross-Entropy Loss

Why does `SoftmaxRegression.forward` stop at the logits instead of returning probabilities? Recall the twin floating-point hazards of computing $\hat{\mathbf{y}} = \mathrm{softmax}(\mathbf{o})$ first and $-\log \hat{y}_y$ second in `float32`:

1. **Overflow:** $\exp(o)$ overflows to `inf` for $o > 88.7$. Subtracting $o_{\max} = \max_k o_k$ from every logit fixes overflow because every shifted exponent $o_k - o_{\max}$ is $\leq 0$.
2. **Underflow:** Even after shifting by $o_{\max}$, any class whose logit sits more than $104$ below $o_{\max}$ (such as class 1 in $\mathbf{o} = (0, -200, 0)$) has $\exp(o_y - o_{\max})$ underflow to exact `0.0` in float32. Taking $\log(0.0)$ then produces `-inf` (or, with our scratch `1e-12` clamp, caps the loss at $27.6$ and **zeroes out the gradient**, leaving the model unable to learn from a confident blunder).

We avoid both overflow and underflow by *never materializing intermediate probabilities at all*. Combining the logarithm and the shifted softmax algebraically via the *LogSumExp trick* turns the division into a subtraction:

$$l(y, \mathbf{o}) = \log \sum_{k=1}^q \exp(o_k) - o_y = o_{\max} + \log \sum_{k=1}^q \exp(o_k - o_{\max}) - o_y.$$

Because every shifted exponent $o_k - o_{\max}$ is $\leq 0$ and at least one equals $0$, the sum inside the logarithm is trapped strictly inside $[1, q]$, and its logarithm lies strictly inside $[0, \log q]$. Neither overflow nor $\log 0$ can occur, and automatic differentiation yields the finite gradient $\mathrm{softmax}(\mathbf{o}) - \mathbf{y}$.

Every framework provides this fused log-sum-exp cross-entropy. We install it once on `d2l.Classifier.loss` so every classifier in the rest of the book inherits it:

```python
@d2l.add_to_class(d2l.Classifier)
def loss(self, Y_hat, Y, averaged=True):
    Y_hat = d2l.reshape(Y_hat, (-1, Y_hat.shape[-1]))
    Y = d2l.reshape(Y, (-1,))
    return F.cross_entropy(
        Y_hat, Y, reduction='mean' if averaged else 'none')
```

<!-- d2l:prose id=softmax-regression-scratch-md-25 fw=all -->

Testing `SoftmaxRegression.loss` on the extreme logits `O = [[1000.0, 0.0, 0.0], [1000.0, 0.0, 0.0]]` with true labels `y = [0, 1]` confirms that the fused LogSumExp implementation returns the exact per-example losses `[0., 1000.]` without overflow or underflow:

```python
O = d2l.tensor([[1000.0, 0.0, 0.0], [1000.0, 0.0, 0.0]])
y = d2l.tensor([0, 1])
SoftmaxRegression(num_outputs=3, lr=0.1).loss(O, y, averaged=False)
```

<!-- d2l:prose id=softmax-regression-scratch-md-26 fw=all -->

(Watch out for one classic pitfall: because `Classifier.loss` expects raw logits and applies log-softmax internally, accidentally placing a softmax at the end of `forward` applies softmax *twice*, trapping the training loss above a positive floor; see Exercise 4 below.)

### Training

Training `SoftmaxRegression` with `d2l.Trainer` takes four lines and reproduces our $\sim 83\%$ validation accuracy with rock-solid numerical stability:

```python
data = d2l.FashionMNIST(batch_size=256)
model = SoftmaxRegression(num_outputs=10, lr=0.1)
trainer = d2l.Trainer(max_epochs=10)
trainer.fit(model, data)
```

## Summary

Implementing softmax regression from scratch introduces the reusable `Classifier` base class, which pairs smooth cross-entropy optimization during training with discrete $\operatorname{argmax}$ accuracy during validation, reaching roughly $83\%$ validation accuracy on Fashion-MNIST in ten epochs. Accumulating one-hot products $\mathbf{P}^\top\mathbf{Y}$ builds the $10 \times 10$ confusion matrix and reveals that linear errors concentrate among the four upper-body garments whose learned $28 \times 28$ weight templates overlap strongly. Because computing `softmax` and `log` separately overflows for large positive logits and underflows to $\log 0$ for large negative logits, our concise implementation returns raw logits from `forward` and evaluates the shifted log-sum-exp cross-entropy $o_{\max} + \log\sum_k \exp(o_k - o_{\max}) - o_y$, guaranteeing finite values and intact gradients across all logit magnitudes.
