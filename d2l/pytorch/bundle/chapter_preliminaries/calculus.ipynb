# Calculus

In Ancient Greece, Archimedes calculated the area of a circle by inscribing regular polygons with increasing numbers of sides $n \to \infty$ (the figure): as the triangles become thinner, their height approaches $r$ and their combined bases approach the circumference $2\pi r$, giving area $\frac{1}{2} r (2\pi r) = \pi r^2$. Approximating curved quantities by simple linear pieces in the limit of small steps lies at the root of both integral calculus and *differential calculus*.

![Regular polygons with 4, 5, 6, and 8 sides inscribed in a circle. As the number of sides grows, the area of the polygon approaches the area of the circle.](../img/polygon-circle.svg)

In deep learning, differential calculus is the engine of *optimization*. Training a model means adjusting its parameters to minimize a scalar loss function measuring how poorly the model fits its training examples. With millions or billions of parameters, probing each parameter by trial and error would require millions of forward passes per update. Instead, the *derivative* gives the instantaneous rate at which the loss changes with respect to each parameter; collecting all partial derivatives into the *gradient* vector identifies the direction of steepest ascent, and the *chain rule* assembles gradients across nested layers in a single backward pass.

In this section, we develop derivatives, gradients, and the chain rule on a running linear-regression example small enough that every number can be verified by hand. In that section, we will see how deep learning frameworks automate these calculations for arbitrary programs.

## Derivatives

### The Derivative as a Limit

Suppose a single-parameter model predicts the constant value $w$ for every example, and we evaluate it on three training examples with labels $1$, $1$, and $0$. Summing the squared errors gives

$$L(w) = (w-1)^2 + (w-1)^2 + (w-0)^2 = 3w^2 - 4w + 2.$$

At $w = 1$, the loss is $L(1) = 1$:

```python
def loss(w):
    return 3 * w ** 2 - 4 * w + 2
```

<!-- d2l:prose id=calculus-md-3 fw=all -->

To see how $L(w)$ responds to a small perturbation $h \neq 0$, we evaluate the difference quotient $\frac{L(1+h) - L(1)}{h}$ as $h$ shrinks from $10^{-1}$ to $10^{-5}$:

```python
for h in 10.0 ** -np.arange(1, 6):
    print(f'h = {h:.0e}   ratio = {(loss(1 + h) - loss(1)) / h:.5f}')
```

<!-- d2l:prose id=calculus-md-4 fw=all -->

Expanding $L(1+h) = 1 + 2h + 3h^2$ shows that $\frac{L(1+h) - L(1)}{h} = 2 + 3h \to 2$ as $h \to 0$. Formally, for $f: \mathbb{R} \rightarrow \mathbb{R}$, the *derivative* of $f$ at $x$ is the limit

$$f'(x) = \lim_{h \rightarrow 0} \frac{f(x+h) - f(x)}{h}.$$

Geometrically, the difference quotient is the slope of the *secant line* through $P = (x, f(x))$ and $Q = (x+h, f(x+h))$; as $h \to 0$, the secant rotates into the *tangent line* at $P$ with slope $f'(x)$ (the figure).

![The difference quotient $(f(x+h)-f(x))/h$ is the slope of the secant through $P$ and $Q$. As $h \to 0$, $Q$ slides toward $P$ and the secant turns into the tangent at $P$, whose slope is $f'(x)$.](../img/calculus-secant-tangent.svg)

Not every function is differentiable everywhere: $|x|$ has a corner at $x = 0$ (slope $-1$ from the left, $+1$ from the right). Even worse, discrete classification accuracy is flat ($f'(x) = 0$) almost everywhere and jumps discontinuously where a prediction flips, providing no gradient signal. That is why classifiers are trained by minimizing smooth surrogate losses such as cross-entropy (the figure).

![The 0--1 error ($1$ for a wrong prediction, $0$ for a correct one) and a smooth surrogate loss, plotted against a score $z$ that is positive when the classifier is right. The 0--1 error is flat on either side of its jump, so its slope says nothing. The surrogate slopes downward everywhere and tells the parameters which way to move.](../img/calculus-surrogate-loss.svg)

### Rules of Differentiation

Given $y = f(x)$, we write $f'(x) = y' = \frac{dy}{dx} = \frac{df}{dx} = \frac{d}{dx} f(x) = Df(x)$. Common elementary derivatives are

$$
\frac{d}{dx} C = 0, \quad \frac{d}{dx} x^n = n x^{n-1}, \quad
\frac{d}{dx} e^x = e^x, \quad \frac{d}{dx} \log x = \frac{1}{x} \quad (x > 0),
$$

and combinations of differentiable functions $f, g$ obey

$$
\begin{aligned}
\frac{d}{dx} [C f(x)] & = C f'(x), &
\frac{d}{dx} [f(x) + g(x)] & = f'(x) + g'(x), \\
\frac{d}{dx} [f(x) g(x)] & = f'(x) g(x) + f(x) g'(x), &
\frac{d}{dx} \frac{f(x)}{g(x)} & = \frac{f'(x) g(x) - f(x) g'(x)}{g(x)^2}.
\end{aligned}
$$

the figure illustrates the product rule geometrically via the area of a $f(x) \times g(x)$ rectangle (full proofs appear in that section). Applying these rules to $L(w) = 3w^2 - 4w + 2$ immediately gives $L'(w) = 6w - 4$, so $L'(1) = 2$.

![The product rule as the area of a rectangle. When $x$ grows by $h$, the sides $f(x)$ and $g(x)$ grow by about $f'(x) h$ and $g'(x) h$, and the area grows by two thin strips and a corner of area $f'(x) g'(x) h^2$, which is negligible for small $h$.](../img/calculus-product-rule.svg)

### First-Order Linear Approximation and Minima

Multiplying $\frac{f(x+h) - f(x)}{h} \approx f'(x)$ by $h$ gives the *first-order linear approximation*

$$f(x + h) \approx f(x) + f'(x)\, h.$$

Because $L'(1) = 2 > 0$, decreasing $w$ lowers the loss. Stepping opposite the derivative via $w \leftarrow w - \eta L'(w)$ with step size $\eta = 0.1$ moves $w$ from $1$ to $1 - 0.1(2) = 0.8$ and drops $L$ from $1$ to $L(0.8) = 0.72$, close to the tangent prediction $1 + 2(-0.2) = 0.60$ (the $0.12$ gap is the neglected quadratic curvature $3h^2$). At any smooth local minimum, the tangent must be horizontal ($f'(x) = 0$); setting $L'(w) = 6w - 4 = 0$ gives the unique global minimizer $w^* = 2/3$, the sample mean of the three labels $1, 1, 0$.

```python
w = np.linspace(0, 2, 101)
d2l.plot(w, [loss(w), 2 * w - 1], 'w', 'L(w)',
         legend=['L(w)', 'tangent at w = 1'])
```

## Partial Derivatives and Gradients

### Partial Derivatives

Now give our three-example problem scalar inputs $0$, $1$, and $2$ paired with labels $1$, $1$, and $0$, and fit an affine line $w_1 + w_2 \cdot \textrm{input}$:

$$L(w_1, w_2) = (w_1 - 1)^2 + (w_1 + w_2 - 1)^2 + (w_1 + 2 w_2 - 0)^2.$$

For a multivariate function $y = f(x_1, \ldots, x_n)$, the *partial derivative* with respect to $x_i$ perturbs only coordinate $i$ while holding all other variables fixed:

$$\frac{\partial f}{\partial x_i} = \lim_{h \rightarrow 0} \frac{f(x_1, \ldots, x_i + h, \ldots, x_n) - f(x_1, \ldots, x_i, \ldots, x_n)}{h}.$$

At $(w_1, w_2) = (1, 0)$, holding $w_2 = 0$ fixed gives $L(w_1, 0) = 3w_1^2 - 4w_1 + 2$, so $\frac{\partial L}{\partial w_1}(1, 0) = 2$; holding $w_1 = 1$ fixed gives $L(1, w_2) = 5w_2^2 + 4w_2 + 1$, with derivative $10w_2 + 4$, so $\frac{\partial L}{\partial w_2}(1, 0) = 4$ (the figure).

![The loss $L(w_1, w_2)$ of the line model near $\mathbf{w} = (1, 0)$ (black dot). Holding $w_2 = 0$ cuts out the orange curve, the loss of the constant model; holding $w_1 = 1$ cuts out the green curve. Their tangents at the dot (dashed) have slopes $\partial L / \partial w_1 = 2$ and $\partial L / \partial w_2 = 4$.](../img/calculus-partial-slices.svg)

In matrix--vector form (that section), writing each example as a row $(1, \textrm{input}_i)$ of $\mathbf{X}$ gives

$$\mathbf{X} = \begin{bmatrix} 1 & 0 \\ 1 & 1 \\ 1 & 2 \end{bmatrix}, \quad \mathbf{y} = \begin{bmatrix} 1 \\ 1 \\ 0 \end{bmatrix}, \quad L(\mathbf{w}) = \|\mathbf{X}\mathbf{w} - \mathbf{y}\|^2.$$

We can check these partial derivatives numerically by defining `line_loss(w)` and a finite-difference helper `numerical_gradient(f, w, h=1e-6)` that perturbs one coordinate of `w` at a time by $h = 10^{-6}$ in `float64`:

```python
X = np.array([[1.0, 0.0], [1.0, 1.0], [1.0, 2.0]])  # rows [1, input]
y = np.array([1.0, 1.0, 0.0])

def line_loss(w):
    return np.sum((X @ w - y) ** 2)

def numerical_gradient(f, w, h=1e-6):
    """Finite differences of f at w, moving one coordinate at a time."""
    grad = np.zeros_like(w)
    for i in range(len(w)):
        step = np.zeros_like(w)
        step[i] = h
        grad[i] = (f(w + step) - f(w)) / h
    return grad
```

<!-- d2l:prose id=calculus-md-6 fw=all -->

Evaluating `numerical_gradient(line_loss, w)` at $\mathbf{w} = (1, 0)$ reproduces our analytical partial derivatives $(2, 4)$ to within the $O(h) \approx 10^{-6}$ truncation error:

```python
w = np.array([1.0, 0.0])
numerical_gradient(line_loss, w)
```

<!-- d2l:prose id=calculus-md-7 fw=all -->

Note that `numerical_gradient` requires $n$ extra evaluations of $f$ for $n$ parameters, which becomes prohibitive when $n = 10^8$.

### Numerical Precision Limits of Finite Differences

Finite differences also suffer from floating-point cancellation when $h$ is too small. In finite precision (machine epsilon $\epsilon \approx 2.2 \times 10^{-16}$ in `float64`, $\epsilon \approx 1.2 \times 10^{-7}$ in `float32`), subtracting nearly equal numbers $f(x+h) - f(x)$ cancels leading digits (*catastrophic cancellation*) and divides the $\sim \epsilon$ rounding residual by $h$. Thus *truncation error* ($\propto h$) and *rounding error* ($\propto \epsilon / h$) compete:

```python
def quotient_error(f, x, df, h):
    """Error of (f(x + h) - f(x)) / h as an estimate of f'(x) = df."""
    return abs((f(x + h) - f(x)) / h - df)

h = np.logspace(-1, -16, 61)  # 61 values of h, evenly spaced on a log scale
```

<!-- d2l:prose id=calculus-md-8 fw=all -->

To compare 64-bit NumPy arithmetic against the 32-bit floating-point format used by default in deep learning, we cast the step grid `h` to a `float32` framework tensor `h32`:

```python
h32 = torch.tensor(h, dtype=torch.float32)
```

<!-- d2l:prose id=calculus-md-9 fw=all -->

We now evaluate `quotient_error(loss, 1, 2, ...)` at $w = 1$ (where the exact derivative is $L'(1) = 2$) across all $61$ step sizes in both `float64` (`h`) and `float32` (`h32`), overlaying the truncation error line $3h$ and the `float32` rounding asymptote $\epsilon / h$:

```python
eps32 = np.finfo(np.float32).eps
d2l.plot(h, [quotient_error(loss, 1, 2, h), quotient_error(loss, 1, 2, h32),
             3 * h, eps32 / h], 'h', 'error', xscale='log', yscale='log',
         ylim=(1e-10, 1e1), legend=['float64', 'float32', '3h', 'ε/h'],
         figsize=(5, 3.5))
```

<!-- d2l:prose id=calculus-md-10 fw=all -->

Balancing $3h \approx \epsilon / h$ puts the optimal step at $h^* \approx \sqrt{\epsilon / 3}$, losing half the format's digits (best error $\sim 10^{-8}$ at $h \approx 10^{-8}$ in `float64`, and $\sim 10^{-4}$ at $h \approx 10^{-4}$ in `float32`). Analytical differentiation via the chain rule avoids both the $O(n)$ cost and the $\sqrt{\epsilon}$ precision loss, leaving `float64` finite differences for unit-test gradient checking.

### The Gradient and Steepest Descent

Concatenating all $n$ partial derivatives of $f: \mathbb{R}^n \to \mathbb{R}$ forms the *gradient* vector $\nabla_{\mathbf{x}} f(\mathbf{x}) = (\frac{\partial f}{\partial x_1}, \ldots, \frac{\partial f}{\partial x_n})^\top$ (matching the shape of $\mathbf{x}$), which generalizes the equation to simultaneous perturbations $\mathbf{h} \in \mathbb{R}^n$:

$$f(\mathbf{x} + \mathbf{h}) \approx f(\mathbf{x}) + \nabla f(\mathbf{x})^\top \mathbf{h}.$$

For a step $\mathbf{h} = t\mathbf{d}$ of length $t > 0$ along unit vector $\mathbf{d}$ ($\|\mathbf{d}\|_2 = 1$), the equation gives $f(\mathbf{x} + t\mathbf{d}) - f(\mathbf{x}) \approx t\,\|\nabla f(\mathbf{x})\|\cos\theta$. Thus $\nabla f(\mathbf{x})$ points in the direction of *steepest ascent* ($\theta = 0^\circ$), $-\nabla f(\mathbf{x})$ points in the direction of *steepest descent* ($\theta = 180^\circ$), and $\nabla f(\mathbf{x})$ is perpendicular to the level set passing through $\mathbf{x}$ ($\theta = 90^\circ$) (the figure).

![Level sets of the loss $L(w_1, w_2)$ of the line model. The gradients (red) are perpendicular to the level set through their base points and point uphill. From $\mathbf{w} = (1, 0)$ (black dot), gradient descent with $\eta = 0.1$ (orange) starts along $-\nabla L$, which does not point at the minimum, and zigzags toward it.](../img/calculus-gradient-field.svg)

This motivates *gradient descent*, $\mathbf{w} \leftarrow \mathbf{w} - \eta \nabla L(\mathbf{w})$: from $\mathbf{w} = (1, 0)$ with learning rate $\eta = 0.1$, one step moves to $(1, 0) - 0.1(2, 4) = (0.8, -0.4)$, cutting the loss from $1.0$ to $0.4$. Why does steepest descent zigzag across the valley in the figure instead of pointing straight at the global minimum $\mathbf{w}^* = (7/6, -1/2)$? Because "steepest" is measured per unit of Euclidean step length $\|\mathbf{h}\|_2$, which depends on the units of each coordinate: stretching one feature axis flattens its contours and tilts $-\nabla L$ away from the minimum, which is why standardizing features (that section) and preconditioning curvature (that section) matter so much in practice.

### Matrix Calculus Identities

Three vector gradient identities (that section) allow us to differentiate vector expressions without writing out coordinates:

$$
\begin{aligned}
\nabla_{\mathbf{x}}\, (\mathbf{a}^\top \mathbf{x}) &= \mathbf{a}, \qquad
\nabla_{\mathbf{x}}\, (\mathbf{x}^\top \mathbf{A} \mathbf{x}) = (\mathbf{A} + \mathbf{A}^\top)\mathbf{x}, \\
\nabla_{\mathbf{x}} \|\mathbf{x}\|^2 &= 2\mathbf{x}.
\end{aligned}
$$

Expanding $L(\mathbf{w}) = \|\mathbf{X}\mathbf{w} - \mathbf{y}\|^2 = \mathbf{w}^\top(\mathbf{X}^\top\mathbf{X})\mathbf{w} - 2(\mathbf{X}^\top\mathbf{y})^\top\mathbf{w} + \mathbf{y}^\top\mathbf{y}$ and applying these rules gives

$$\nabla L(\mathbf{w}) = 2\mathbf{X}^\top \mathbf{X} \mathbf{w} - 2\mathbf{X}^\top \mathbf{y} = 2\mathbf{X}^\top (\mathbf{X}\mathbf{w} - \mathbf{y}).$$

In words, each component of the gradient is twice the dot product of the corresponding feature column of $\mathbf{X}$ with the residual vector $\mathbf{r} = \mathbf{X}\mathbf{w} - \mathbf{y}$; at the optimum $\nabla L(\mathbf{w}^*) = \mathbf{0}$, the prediction errors are orthogonal to every feature.

## The Chain Rule

Deep networks compose many nested layers $y = f_L(f_{L-1}(\cdots f_1(\mathbf{x})\cdots))$; differentiating compositions without expanding them relies on the *chain rule*.

For scalar functions $z = f(u)$ and $u = g(x)$, rates of change multiply: $\frac{dz}{dx} = \frac{dz}{du}\frac{du}{dx}$. More generally, if $z = f(u_1, \ldots, u_m)$ where each $u_j = g_j(x_1, \ldots, x_n)$, rates multiply along each path $x_i \to u_j \to z$ and sum across all parallel paths:

$$\frac{\partial z}{\partial x_i} = \sum_{j=1}^m \frac{\partial z}{\partial u_j} \frac{\partial u_j}{\partial x_i} \quad \Longrightarrow \quad \nabla_{\mathbf{x}} z = \mathbf{J}^\top \nabla_{\mathbf{u}} z,$$

where $\mathbf{J} \in \mathbb{R}^{m \times n}$ is the *Jacobian matrix* of $\mathbf{u}$ with respect to $\mathbf{x}$, with entries $J_{ji} = \frac{\partial u_j}{\partial x_i}$. For an affine map $\mathbf{u} = \mathbf{A}\mathbf{x} - \mathbf{y}$, the Jacobian is simply $\mathbf{J} = \mathbf{A}$.

A small case is worth working by hand. Let $z = u^2 + v$ with $u = x_1 x_2$ and $v = x_2^2$, evaluated at $(x_1, x_2) = (1, 2)$. The input $x_2$ reaches $z$ along two paths, one through $u$ and one through $v$ (the figure). Each path contributes the product of its local derivatives, and the two contributions add: $\partial z / \partial x_2 = 4 \cdot 1 + 1 \cdot 4 = 8$.

![The chain rule for $z = u^2 + v$ with $u = x_1 x_2$ and $v = x_2^2$ at $(x_1, x_2) = (1, 2)$. Each box shows the value of a variable (black) and the derivative of $z$ with respect to it (red), and each red arrow carries a local derivative. Derivatives multiply along a path and add where two paths meet, as they do at $x_2$.](../img/calculus-parallel-dag.svg)

Decomposing our line loss $L(\mathbf{w}) = \|\mathbf{X}\mathbf{w} - \mathbf{y}\|^2$ into two stages, namely (1) inner residuals $\mathbf{r} = \mathbf{X}\mathbf{w} - \mathbf{y}$ with Jacobian $\mathbf{J} = \mathbf{X}$ and (2) outer squared norm $L = \|\mathbf{r}\|^2$ with gradient $\nabla_{\mathbf{r}} L = 2\mathbf{r}$, the multivariate chain rule recovers $\nabla_{\mathbf{w}} L = \mathbf{X}^\top(2\mathbf{r})$ in two passes:

```python
r = X @ w - y    # forward: errors, from which the loss is r @ r
grad_r = 2 * r   # backward: gradient of the loss with respect to r
X.T @ grad_r     # backward: gradient with respect to w
```

<!-- d2l:prose id=calculus-md-11 fw=all -->

The *forward pass* computes $\mathbf{r} = \mathbf{X}\mathbf{w} - \mathbf{y}$ and $L = \mathbf{r}^\top\mathbf{r}$; the *backward pass* (*backpropagation*) multiplies $2\mathbf{r}$ by $\mathbf{X}^\top$, computing the exact gradient with respect to all parameters at roughly the cost of one forward evaluation. In that section, we hand this mechanical backward pass over to the framework's automatic differentiation engine.

## Summary

The derivative $f'(x) = \lim_{h \to 0} \frac{f(x+h) - f(x)}{h}$ gives the slope of the tangent line and the local linear approximation $f(x+h) \approx f(x) + f'(x)h$. For $f: \mathbb{R}^n \to \mathbb{R}$, the gradient $\nabla_{\mathbf{x}} f(\mathbf{x})$ collects all partial derivatives, satisfies $f(\mathbf{x}+\mathbf{h}) \approx f(\mathbf{x}) + \nabla f(\mathbf{x})^\top \mathbf{h}$, and points in the direction of steepest ascent (perpendicular to level sets), motivating gradient descent $\mathbf{w} \leftarrow \mathbf{w} - \eta \nabla L(\mathbf{w})$. Unlike finite differences, which cost $O(n)$ evaluations and lose precision to catastrophic cancellation ($\sim \sqrt{\epsilon}$), the multivariate chain rule $\nabla_{\mathbf{x}} z = \mathbf{J}^\top \nabla_{\mathbf{u}} z$ propagates exact gradients backward from the loss to all parameters in a single backward pass (*backpropagation*).
