# Practice

In our small experiments so far, training took a few seconds per run, allowing us to sweep grids of learning rates and repeat runs whenever an update diverged. Production training runs are different: when a single training run takes days or weeks, most settings must be fixed in advance, a single unguarded gradient spike can corrupt hours of progress, and the final checkpoint must not depend on the luck of the last minibatch draw.

Even at extreme scale (such as the $540$-billion-parameter PaLM model trained across $6{,}144$ TPU chips, which experienced roughly $20$ sudden loss spikes despite active gradient clipping [@chowdhery2022palm]), reliable training rests on three operational pillars:

1. **Guards against transient instabilities:** Global-norm **gradient clipping** (that section) caps anomalous steps under SGD and protects AdamW's second-moment accumulator $\mathbf{v}_t$ from transient gradient spikes, paired with checkpoint-and-rewind recovery.
2. **Iterate smoothing:** Maintaining an **exponential moving average (EMA)** of model weights eliminates high-frequency noise-ball fluctuations without requiring a full learning-rate decay to zero.
3. **Systematic calibration from short proxy runs:** Selecting the batch size from device throughput and $b_{\text{noise}}$ (that section), adopting battle-tested optimizer defaults, and calibrating the peak learning rate on a short run ($10\%$ of the horizon) before executing the full warmup-and-decay schedule (that section).

In this section, we examine each of these three pillars and assemble them into a complete end-to-end configuration for an untuned $3.3$-million-parameter, $4$-block `TinyLM`.

## Gradient Clipping and Spike Recovery

As introduced for recurrent networks in that section, global-norm **gradient clipping** treats the concatenated gradient across all parameters as a single vector $\mathbf{g}$ and rescales it whenever its Euclidean norm $\|\mathbf{g}\|_2$ exceeds a threshold $\theta$:

$$
\mathbf{g} \leftarrow \min\left(1,\; \frac{\theta}{\|\mathbf{g}\|_2}\right) \mathbf{g}.
$$

Because the scaling factor $\theta / \|\mathbf{g}\|_2$ is global, clipping preserves the relative direction across all layers while capping the Euclidean norm of the gradient entering the optimizer at $\theta$.

```python
class Clipped:
    """Clip the global gradient norm before every optimizer step."""
    def __init__(self, optimizer, params, max_norm=1.0):
        self.optimizer, self.params = optimizer, list(params)
        self.max_norm, self.norms = max_norm, []

    def step(self):
        self.norms.append(nn.utils.clip_grad_norm_(self.params, self.max_norm))
        self.optimizer.step()

    def zero_grad(self):
        self.optimizer.zero_grad()
```

### Clipping as a Guard vs. a Constant Brake

In that section, SGD with momentum $0.9$ on `TinyLM` diverged to `inf` at $\eta = 1.0$ within the first $20$ steps. Is $\eta = 1.0$ too large for the entire loss surface, or only during the initial transient steps where freshly initialized weights produce large gradients? We first load the `WikiText103` stream ($b = 64$, window length $64$):

```python
data = d2l.WikiText103(batch_size=64, num_steps=64)
```

<!-- d2l:prose id=practice-md-4 fw=all -->

Next, `run(seed, lr, max_norm)` trains `TinyLM` for $2{,}000$ steps of SGD with momentum $0.9$ clipped at `max_norm`, returning both the step losses and the pre-clip gradient norms:

```python
new_seed = lambda: random.getrandbits(31)

def run(seed, lr, max_norm, num_steps=2000):
    """SGD with momentum 0.9 on TinyLM, clipped at max_norm (math.inf: not
    at all); return the losses and the gradient norms before clipping."""
    torch.manual_seed(seed)  # The initialization and the minibatch order
    model = d2l.TinyLM(len(data.vocab))
    optimizer = Clipped(torch.optim.SGD(model.parameters(), lr, momentum=0.9),
                        model.parameters(), max_norm)
    losses = d2l.train_steps(model, data, optimizer, num_steps)
    return losses, torch.stack(optimizer.norms).cpu().numpy()
```

<!-- d2l:prose id=practice-md-5 fw=all -->

We now test $(\eta, \theta) \in \{(1.0, \infty), (1.0, 1.0), (0.3, \infty), (3.0, 1.0)\}$ across two seeds, logging how many steps trigger clipping ($\|\mathbf{g}_t\|_2 > \theta$), the last step on which clipping fires, and the median gradient norm:

```python
settings = [(1.0, math.inf), (1.0, 1.0), (0.3, math.inf), (3.0, 1.0)]
seeds, runs = [new_seed() for _ in range(2)], {}
print(f'{"lr":>4}{"clip at":>9}{"final loss":>12}{"steps clipped":>15}'
      f'{"last":>6}{"median norm":>13}{"largest / median":>18}')
for seed in seeds:
    for lr, max_norm in settings:
        losses, norms = runs[seed, lr, max_norm] = run(seed, lr, max_norm)
        line = f'{lr:4g}{max_norm:9g}{d2l.final_loss(losses):12.3f}'
        if max_norm < math.inf:
            fired = np.nonzero(norms > max_norm)[0]  # Steps clipping changed
            median = np.median(norms)
            line += (f'{len(fired):15d}{fired.max(initial=0):6d}{median:13.2f}'
                     f'{norms[100:].max() / median:18.1f}')  # After step 100
        print(line)
```

<!-- d2l:prose id=practice-md-6 fw=all -->

At $\eta = 1.0$, clipping at $\theta = 1.0$ fires on only the first few dozen steps (all before step $50$) and never fires again over the remaining $1{,}950$ steps (where the median gradient norm is $\sim 0.25$), rescuing the run from `inf` down to $\sim 1.61\text{--}1.64$ nats. Here clipping acts as a **guard**: it intervenes on $<2\%$ of steps to prevent early transient blow-up (and handle heavy-tailed gradient noise [@Zhang.Karimireddy.Veit.ea.2020]). By contrast, at $\eta = 3.0$, clipping fires on every step for as long as the gradient norm stays finite, and the run still fails to train (its loss stalls at $100\text{--}200$ nats or overflows to `inf`): when clipping is permanently active as a **brake**, the learning rate $\eta$ is simply too large [@Godbole.Dahl.Gilmer.ea.2023]. In practice, set $\theta$ roughly $3\text{--}5\times$ above the steady-state median gradient norm (typically $\theta = 1.0$) and monitor the clipping fraction.

### Clipping Under AdamW and Operational Spike Recovery

Under AdamW (that section), a single anomalous gradient $g \gg \sqrt{v_{t-1}}$ already produces a bounded step of at most $\eta(1-\beta_1)/\sqrt{1-\beta_2}$ ($\approx 3.2\,\eta$ at $\beta_2 = 0.999$, or $0.45\,\eta$ at $\beta_2 = 0.95$). Instead of bounding the immediate step, **clipping under AdamW protects the second-moment state $\mathbf{v}_t$**: an unclipped $100\times$ gradient spike injects $(1-\beta_2) \times 10^4$ times the normal squared gradient into $v_t$, inflating the denominator $\sqrt{\hat{v}_t}$ and paralyzing learning for the next $\sim 1/(1-\beta_2)$ steps ($1{,}000$ steps at $\beta_2 = 0.999$; $20$ steps at $\beta_2 = 0.95$).

When a sudden loss spike occurs despite gradient clipping (and auxiliary logit regularizers such as PaLM's $z$-loss $10^{-4}\log^2(\sum_j e^{o_j})$; Exercise 2), the standard operational remedy [@chowdhery2022palm; @zhang2022opt] is to **rewind to a checkpoint saved $\sim 100$ steps prior to the spike and skip the $200\text{--}500$ minibatches surrounding the trigger**, which requires checkpointing the model weights, optimizer moments $(\mathbf{m}_t, \mathbf{v}_t)$, scheduler step counter, and data-stream cursor at regular intervals.

## Weight Averaging (EMA)

At a constant or non-zero learning rate, stochastic iterates fluctuate continuously around the optimum inside the noise ball (that section), where a mode of curvature $\lambda_{\text{curv}}$ has autocorrelation time $\sim 1/(\eta\lambda_{\text{curv}})$ steps. Maintaining an **exponential moving average (EMA)** $\bar{\mathbf{x}}_t$ of the live parameters $\mathbf{x}_t$ with decay coefficient $\alpha \in [0.99, 0.9999]$ [@Polyak.Juditsky.1992],

$$
\bar{\mathbf{x}}_t = \alpha\, \bar{\mathbf{x}}_{t-1} + (1 - \alpha)\, \mathbf{x}_t,
$$

averages over a trailing window of $1/(1-\alpha)$ steps to cancel noise-ball fluctuations at evaluation time without altering the live training trajectory.

We compare constant-rate SGD ($\eta = 0.3$) with and without weight EMA ($\alpha = 0.999$, window $1{,}000$ steps $\approx 4.2$ epochs, started after epoch $5$ to avoid initial lag) against cosine learning-rate decay over $15$ epochs ($3{,}525$ steps) on `FashionCNN`. First we load `FashionMNIST` ($b = 256$) and define the validation accuracy helper `test_acc`:

```python
fashion = d2l.FashionMNIST(batch_size=256)
num_epochs, num_batches = 15, len(fashion.train_dataloader())
test_acc = lambda model: d2l.evaluate_accuracy(model, fashion.val_dataloader())
```

<!-- d2l:prose id=practice-md-7 fw=all -->

Next we define the `warmup_cosine` schedule helper and a weight EMA:

<!-- d2l:prose id=practice-md-8 fw=pytorch -->

PyTorch's `torch.optim.swa_utils.AveragedModel` keeps the EMA, so only the schedule needs defining here.

```python
def warmup_cosine(peak, warmup, num_steps):
    """A linear warmup to peak over warmup steps, then a cosine to zero."""
    def schedule(t):
        if t < warmup:
            return peak * t / warmup
        return peak * (1 + math.cos(math.pi * (t - warmup)
                                    / (num_steps - warmup))) / 2
    return schedule
```

<!-- d2l:prose id=practice-md-9 fw=all -->

The training loop `train_fashion` runs plain SGD under `schedule(t)` for `num_epochs=15` epochs, recording held-out test accuracy at the end of every epoch for both the live weights and their EMA (started at epoch `start=5`):

```python
def train_fashion(seed, schedule, alpha=0.999, start=5):
    """Plain SGD on FashionCNN at the rate schedule(t); the test accuracy
    after every epoch of the weights and of their EMA from epoch start on."""
    torch.manual_seed(seed)  # The initialization and the minibatch order
    model = d2l.FashionCNN().to(d2l.try_gpu())
    optimizer = torch.optim.SGD(model.parameters(), lr=1.0)  # Times schedule
    scheduler = torch.optim.lr_scheduler.LambdaLR(optimizer, schedule)
    optimizer.register_step_post_hook(lambda *_: scheduler.step())
    ema = torch.optim.swa_utils.AveragedModel(
        model, multi_avg_fn=torch.optim.swa_utils.get_ema_multi_avg_fn(alpha))
    live, avg = [], []
    for epoch in range(num_epochs):
        if epoch == start:  # Its first update copies the current weights
            optimizer.register_step_post_hook(
                lambda *_: ema.update_parameters(model))
        d2l.train_steps(model, fashion, optimizer, num_batches)
        live.append(test_acc(model))
        avg.append(test_acc(ema) if epoch >= start else math.nan)
    return live, avg
```

<!-- d2l:prose id=practice-md-10 fw=all -->

We run `train_fashion` across two seeds for both a constant learning rate ($\eta = 0.3$) and cosine decay ($\eta: 0.3 \to 0$), comparing the range of live accuracies over the last five epochs against the final EMA accuracy:

```python
schedules = {'constant rate': lambda t: 0.3,
             'cosine decay': warmup_cosine(0.3, 0, num_epochs * num_batches)}
cnn_seeds, curves = [new_seed() for _ in range(2)], {}
print(f'{"test accuracy":<15}{"live, last five epochs":>24}{"live, final":>13}'
      f'{"EMA, final":>12}')
for seed in cnn_seeds:
    for name, schedule in schedules.items():
        live, avg = curves[seed, name] = train_fashion(seed, schedule)
        print(f'{name:<15}{min(live[-5:]):15.3f} to {max(live[-5:]):.3f}'
              f'{live[-1]:13.3f}{avg[-1]:12.3f}')
```

<!-- d2l:prose id=practice-md-11 fw=all -->

Plotting the epoch-by-epoch test accuracy trajectories for the first seed visualizes how the constant-rate EMA eliminates the noise-ball jitter of the live weights:

```python
(live, avg), (decayed, _) = (curves[cnn_seeds[0], name] for name in schedules)
d2l.plot(list(range(1, num_epochs + 1)), [live, avg, decayed], 'epoch',
         'test accuracy', ylim=[0.85, 0.925], figsize=(5, 2.5),
         legend=['constant rate, live weights', 'constant rate, EMA',
                 'cosine decay, live weights'], legend_loc='outside')
```

<!-- d2l:prose id=practice-md-12 fw=all -->

At a constant learning rate, the live weights jitter over a test-accuracy band of a few percentage points across the last five epochs, and occasionally far more when a single epoch goes badly, whereas their EMA rises smoothly along the top of the band to about $91\%$—matching the final accuracy of the cosine-decayed run without ever lowering the learning rate. Thus, during the constant plateau of a WSD schedule, weight EMA is a cheap approximation to a cooldown branch (that section): it costs no extra steps. It is not a full replacement. On language models, averaging along the plateau recovers part of the cooldown gain, not all of it [@Hagele.Bakouch.Kosson.ea.2024].

Following the table, we configure our $3.3\text{M}$-parameter `TinyLM` with:

1. **Batch size:** $b = 256$ sequences (where hardware throughput saturates and comparable to the post-warmup gradient-noise scale $b_{\text{noise}} \sim 100\text{--}200$).
2. **Optimizer & guards:** AdamW with $(\beta_1, \beta_2) = (0.9, 0.95)$, decoupled weight decay $\lambda = 0.1$ applied exclusively to 2D hidden matrices (that section), global-norm gradient clipping at $\theta = 1.0$, and a short-window weight EMA ($\alpha = 0.995$).
3. **Peak learning rate calibration:** Three $300$-step ($10\%$-horizon) trial runs across $\eta \in \{10^{-3}, 3 \times 10^{-3}, 10^{-2}\}$ identify $\eta_0 = 3 \times 10^{-3}$ as optimal (with $<5\%$ of steps clipped, confirming that $\theta = 1.0$ acts purely as a guard).
4. **Schedule:** $60$-step linear warmup ($2\%$ of $T = 3{,}000$) followed by cosine decay to zero, which lowers validation loss by $0.05\text{--}0.06$ nats below the constant-rate baseline ($<0.01$-nat seed spread).

We first define `tiny_lm(seed)` to instantiate the $4$-block, width-$256$, $4$-head `TinyLM` and `sample_lm(b)` to draw random training windows for noise-scale estimation:

```python
def tiny_lm(seed):
    """A TinyLM of four blocks of width 256 on the GPU."""
    torch.manual_seed(seed)  # The initialization and the minibatch order
    return d2l.TinyLM(len(data.vocab), d_model=256, num_heads=4,
                      num_blks=4).to(d2l.try_gpu())

def sample_lm(b):  # b windows at random offsets in the training text
    starts = torch.randint(len(data.train) - data.num_steps, (b,))
    X, Y = data.windows(data.train, starts.numpy())
    return X.to(d2l.try_gpu()), Y.to(d2l.try_gpu())
```

<!-- d2l:prose id=practice-md-13 fw=all -->

The harness `train(seed, schedule, num_steps)` configures AdamW with $(\beta_1, \beta_2) = (0.9, 0.95)$ and weight decay $0.1$ restricted to 2D non-embedding matrices, wraps it in global-norm clipping (`max_norm=1.0`) and weight EMA (`alpha=0.995`), and returns the losses, live model, EMA model, and clipping rate:

```python
def train(seed, schedule, num_steps, max_norm=1.0, alpha=0.995):
    """AdamW with betas (0.9, 0.95) and weight decay 0.1 on the matrices at
    the rate schedule(t), clipped at max_norm, with an EMA of the weights;
    return the losses, the model, the EMA and the fraction of clipped steps."""
    model = tiny_lm(seed)
    matrices = [p for name, p in model.named_parameters()
                if p.ndim == 2 and 'emb' not in name]
    others = [p for name, p in model.named_parameters()
              if p.ndim != 2 or 'emb' in name]
    adamw = torch.optim.AdamW([{'params': matrices, 'weight_decay': 0.1},
                               {'params': others, 'weight_decay': 0.0}],
                              lr=1.0, betas=(0.9, 0.95))  # Times schedule
    scheduler = torch.optim.lr_scheduler.LambdaLR(adamw, schedule)
    ema = torch.optim.swa_utils.AveragedModel(
        model, multi_avg_fn=torch.optim.swa_utils.get_ema_multi_avg_fn(alpha))
    adamw.register_step_post_hook(lambda *_: scheduler.step())
    adamw.register_step_post_hook(lambda *_: ema.update_parameters(model))
    optimizer = Clipped(adamw, model.parameters(), max_norm)
    losses = d2l.train_steps(model, data, optimizer, num_steps)
    clip_rate = (torch.stack(optimizer.norms) > max_norm).float().mean()
    return losses, model, ema, float(clip_rate)
```

<!-- d2l:prose id=practice-md-14 fw=all -->

At $b = 256$, we measure `d2l.noise_scale` at initialization, run three $300$-step ($10\%$-horizon) proxy trials across $\eta \in \{10^{-3}, 3 \times 10^{-3}, 10^{-2}\}$ to select the peak learning rate `peak`, and re-estimate `noise_scale` after $300$ steps:

```python
data.batch_size = 256  # Where the throughput levels off
seed, short = new_seed(), {}
print('noise scale at initialization: '
      f'{d2l.noise_scale(tiny_lm(seed), sample_lm, b_small=8, b_big=256):.0f}')
for lr in [1e-3, 3e-3, 1e-2]:
    losses, model, _, clip_rate = train(seed, lambda t: lr, num_steps=300)
    short[lr] = (d2l.final_loss(losses), model)
    print(f'lr {lr:g}: loss {short[lr][0]:.3f} after 300 steps, '
          f'{clip_rate:.0%} of the steps clipped')
peak = min(short, key=lambda lr: short[lr][0])
print(f'peak rate {peak:g}; noise scale after its 300 steps: '
      f'{d2l.noise_scale(short[peak][1], sample_lm, b_small=8, b_big=256):.0f}')
```

<!-- d2l:prose id=practice-md-15 fw=all -->

Finally, we execute the full $3{,}000$-step run across two seeds comparing our configured schedule (`warmup_cosine(peak, 60, 3000)`) and its weight EMA against a constant-rate baseline at `peak`:

```python
num_steps = 3000
val_loss = lambda model: math.log(d2l.lm_perplexity(model, data.val_dataloader()))
print(f'{"validation loss":<16}{"configured":>11}{"its EMA":>9}'
      f'{"clipped":>9}{"baseline":>10}{"gap":>7}')
results = []
for seed in [new_seed() for _ in range(2)]:
    _, model, ema, clip_rate = train(
        seed, warmup_cosine(peak, num_steps // 50, num_steps), num_steps)
    _, baseline, _, _ = train(seed, lambda t: peak, num_steps)
    results.append((val_loss(model), val_loss(baseline)))
    print(f'{"seed " + str(len(results)):<16}{results[-1][0]:11.3f}'
          f'{val_loss(ema):9.3f}{clip_rate:9.1%}{results[-1][1]:10.3f}'
          f'{results[-1][1] - results[-1][0]:7.3f}')
(c1, b1), (c2, b2) = results
print(f'spread between the seeds: configured {abs(c1 - c2):.3f}, '
      f'baseline {abs(b1 - b2):.3f}')
```

## Summary

**Global-norm gradient clipping** ($\mathbf{g} \leftarrow \min(1, \theta/\|\mathbf{g}\|_2)\mathbf{g}$) acts as a **guard** when it fires on a small fraction ($<5\%$) of early or anomalous steps, preventing initial blow-up under SGD and protecting AdamW's second-moment buffer $\mathbf{v}_t$ from gradient spikes, whereas continuous firing signals that $\eta$ is too large and clipping has become a **brake**. **Exponential moving averaging (EMA)** of model weights ($\bar{\mathbf{x}}_t = \alpha\bar{\mathbf{x}}_{t-1} + (1-\alpha)\mathbf{x}_t$) smooths out stationary noise-ball fluctuations without modifying the live trajectory, matching the accuracy of a full cosine decay on constant-rate runs. Reliable large-scale training combines hardware-saturated batch sizes positioned near $b_{\text{noise}}$, AdamW with $(\beta_1, \beta_2) = (0.9, 0.95)$ and decoupled weight decay $\lambda = 0.1$ restricted to 2D weight matrices, global-norm clipping at $\theta = 1.0$, short-horizon peak-LR calibration, and linear warmup followed by cosine or WSD cooldown.
